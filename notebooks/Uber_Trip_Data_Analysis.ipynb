
# Uber Trip Data Analysis

## 1. Introduction

Uber is a ride-hailing platform that connects passengers with drivers.

This project analyzes Uber pickup data to understand ride demand patterns, peak hours, daily trends, and geographical distribution.

## 2. Business Problem

Understanding ride demand is important for efficient driver allocation and operational planning.

This project aims to identify when and where ride demand is highest using historical Uber pickup data.

## 3. Project Objectives

- Analyze hourly ride demand.
- Identify the busiest days.
- Understand pickup patterns.
- Identify high-demand geographical locations.
- Discover useful business insights.

```python
df_clean=df.drop_duplicates()
df_clean.shape
```

```python
df_clean.duplicated().sum()
```

```python
df_clean['Date/Time']=pd.to_datetime(df_clean['Date/Time'])
df_clean.info()
```

```python
df_clean['Hour']=df_clean['Date/Time'].dt.hour
df_clean[['Date/Time','Hour']].head()
```

```python
df_clean['Day']=df_clean['Date/Time'].dt.day
df_clean[['Date/Time','Day']].head()
```

```python
df_clean['Month']=df_clean['Date/Time'].dt.month
df_clean[['Date/Time','Month']].head()
```

```python
df_clean['DayOfWeek']=df_clean['Date/Time'].dt.day_name()
df_clean[['Date/Time','DayOfWeek']].head()
```

```python
df_clean.head()
```

```python
hourly_pickups=df_clean['Hour'].value_counts().sort_index()
hourly_pickups
```

```python
df_clean.to_csv("../dataset/uber_cleaned.csv", index=False)
```
